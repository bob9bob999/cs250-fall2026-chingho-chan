```python
sentence = "the quick brown fox jumps over the lazy dog the end"

words = sentence.split()  # .split() breaks down the sentence into a list of words

# 1. Count each word
counts = {}  # create an empty dictionary

for word in words:      # for loop to check words one by one
    if word in counts:    # already exist → count +1
        counts[word] += 1
    else:                 # not yet exist → add to dict. with count 1
        counts[word] = 1

print(counts)


# 2. Find the longest word
longest = ""  # empty str

for word in words:      # for loop to check words one by one
    if len(word) > len(longest):
        longest = word

print("Longest word:", longest)

    # in case of tie, show the first longest word (quick)
    # although "brown" and "jumps" have 5 letters, they don't trigger the "if" condition,
    # because 5 > 5 is false

# 3. Title-case the sentence
title_sentence = " ".join(
    [word[0].upper() + word[1:] for word in words]
)

print(title_sentence)

    # This takes the first character, capitalizes it using . upper(),
    # then adds the rest of the word past the first character back to the word, seperated by a space.
```
